## Simple LLM for Thai Language

This notebook demonstrates how to use a pre-trained Large Language Model (LLM) that supports the Thai language. We will use the `transformers` library from Hugging Face to load a model and tokenizer, and then generate text.

### 1. Install Dependencies

### 2. Import Libraries and Load Model
Next, we'll import the necessary classes from `transformers` and load a pre-trained model and its tokenizer. For Thai text generation, we will use `facebook/xglm-564M`. This is a multilingual Causal Language Model that supports the Thai language.

```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

# Check if GPU is available and set device
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f"Using device: {device}")

# Model name for Thai language (XGLM supports Thai generation)
model_name = "facebook/xglm-564M"

# Load tokenizer and model
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name).to(device)

print(f"Model '{model_name}' loaded successfully!")
```

### 3. Define Text Generation Function
We'll create a helper function to make text generation easier. This function will take a prompt and some generation parameters, then return the generated text.

```python
def generate_thai_text(prompt, max_length=100, num_return_sequences=1, temperature=0.7, top_k=50, top_p=0.95):
    # Encode the input prompt
    input_ids = tokenizer.encode(prompt, return_tensors="pt").to(device)

    # Generate text
    output = model.generate(
        input_ids,
        max_length=max_length,
        num_return_sequences=num_return_sequences,
        temperature=temperature,
        top_k=top_k,
        top_p=top_p,
        do_sample=True, # เปิดใช้งานโหมด Sampling เพื่อให้ค่า temperature ทำงาน
        pad_token_id=tokenizer.eos_token_id # Important for generation
    )

    # Decode the generated text
    generated_texts = [tokenizer.decode(ids, skip_special_tokens=True) for ids in output]
    return generated_texts
```

### 4. Demonstrate Usage
Now, let's try generating some text with Thai prompts. Feel free to modify the prompt and parameters to experiment.

```python
# Example 1: Simple completion
prompt = "สวัสดีครับ วันนี้อากาศดี"
print(f"Prompt: {prompt}")
generated_text = generate_thai_text(prompt, max_length=50)
for i, text in enumerate(generated_text):
    print(f"Generated Text {i+1}: {text}")

print("\n" + "-"*30 + "\n")

```

```python
# Example 2: More creative prompt
prompt = "กาลครั้งหนึ่งนานมาแล้ว มีกระต่ายตัวหนึ่ง"
print(f"Prompt: {prompt}")
generated_text = generate_thai_text(prompt, max_length=150, temperature=0.8)
for i, text in enumerate(generated_text):
    print(f"Generated Text {i+1}: {text}")
```
